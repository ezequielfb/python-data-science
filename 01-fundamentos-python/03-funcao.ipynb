```python
def maior(x, y, z):
    if x >= y and x >= z:
        return x
    elif y >= z:
        return y
    
    return z 
```

```python
def calcular_preco(preco:float, desconto: int= 0) -> float:
    valor_final = preco - (preco * desconto / 100)
    return valor_final
```
