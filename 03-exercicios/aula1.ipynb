Exercício 1:   
Desenvolva um programa que selecione um número aleatório entre 0 e 10. O usuário deverá tentar adivinhar esse número. Ao final, o programa exibirá a sequência de palpites do usuário e o total de tentativas feitas até acertar o número correto.

```python
import random
NumeroGerado: int = random.randint(0, 10)

print("Bem-vindo ao jogo de adivinhação!")
print("Tente adivinhar o número entre 0 e 10.")
print("Você tem 5 tentativas para acertar o número.")
```

```python
from sqlalchemy import true


NumeroUsuario: int

Palpites = [] #lista para armazenar os palpites do usuário

Palpites = []

for tentativa in range(5):
    NumeroUsuario = int(input("Digite seu palpite: "))
    Palpites.append(NumeroUsuario)

    if NumeroUsuario == NumeroGerado:
        print("Parabéns! Você acertou!")
        break
    elif NumeroUsuario > NumeroGerado:
        print("O número que você digitou é maior.")
    else:
        print("O número que você digitou é menor.")
else:
    print("Você não acertou.")
    
for palpite in Palpites:
    print(f"Você digitou o número: {palpite}")
```

Exercício 2:   
Você está organizando uma lista de alunos por turmas em uma escola. Cada turma contém vários alunos, e cada aluno é representado por um dicionário com informações sobre seu nome e idade. 

a) Criar as turmas: Crie duas turmas, cada uma contendo uma lista de dicionários representando os alunos. Cada dicionário deve ter as chaves "nome" e "idade".   
b) Adicionar um novo aluno: Adicione um novo aluno em uma das turmas.   
c) Remover um aluno: Remova um aluno de uma das turmas.   
d) Exibir os alunos de uma turma

```python
turmas = {
    "Turma A": [
        {"nome": "João", "idade": 20, "id": 1},
        {"nome": "Maria", "idade": 22, "id": 2},
        {"nome": "Pedro", "idade": 21, "id": 3}
    ],
    "Turma B": [
        {"nome": "Ana", "idade": 23, "id": 4},
        {"nome": "Lucas", "idade": 24, "id": 5},
        {"nome": "Carla", "idade": 25, "id": 6}
    ]
}

```

```python
NovoAluno = {"nome": "Rafael", "idade": 26, "id": 7}

turmas["Turma A"].append(NovoAluno)

print(turmas)
```

```python
id: int = 3

print(turmas["Turma A"])
print(turmas["Turma B"], end="\n\n\n")

for turma in turmas.keys():
    for aluno in turmas[turma]:
        if aluno["id"] == id:
            turmas[turma].remove(aluno)
            
print(turmas["Turma A"])
print(turmas["Turma B"])
```

```python
for turma in turmas.keys():
    print(turma)
    
    print("Alunos: ")
    for aluno in turmas[turma]:
        print(f"Nome: {aluno['nome']}, Idade: {aluno['idade']}, ID: {aluno['id']}")
    print("\n")
```
